```python
# Import libraries:

import pandas as pd
import numpy as np
import os
```

```python
# Load the dataset:

country = pd.read_csv("dataset_raw/country_ai_index.csv")
company = pd.read_csv("dataset_raw/ai_company_adoption.csv")
industry = pd.read_csv("dataset_raw/ai_industry_summary.csv")
```

```python
# Dataset size:

print("Country dataset:", country.shape)
print("Company dataset:", company.shape)
print("Industry dataset:", industry.shape)
```

```python
print("COUNTRY DATASET")
country.info()

print("\nCOMPANY DATASET")
company.info()

print("\nINDUSTRY DATASET")
industry.info()
```

```python
print("Country missing values:")
print(country.isnull().sum())

print("\nCompany missing values:")
print(company.isnull().sum())

print("\nIndustry missing values:")
print(industry.isnull().sum())
```

```python
print("Country duplicates:", country.duplicated().sum())
print("Company duplicates:", company.duplicated().sum())
print("Industry duplicates:", industry.duplicated().sum())
```

```python
# Country relationships: all company countries exist in the country table

company_countries = set(company["country"].unique())
country_reference = set(country["country"].unique())

print("Company countries not in country dataset:",
      company_countries - country_reference)

print("Country dataset countries not in company dataset:",
      country_reference - company_countries)
```

```python
# Industry relationships: all company industries exist in the industry table

company_industries = set(company["industry"].unique())
industry_reference = set(industry["industry"].unique())

print("Company industries not in industry dataset:",
      company_industries - industry_reference)

print("Industry dataset industries not in company dataset:",
      industry_reference - company_industries)
```

```python
country_clean = country.copy()
company_clean = company.copy()
industry_clean = industry.copy()
```

```python
for df in [country_clean, company_clean, industry_clean]:
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )
```

```python
country_clean.to_csv(
    "dataset_cleaned/country_ai_index_clean.csv",
    index=False
)

company_clean.to_csv(
    "dataset_cleaned/ai_company_adoption_clean.csv",
    index=False
)

industry_clean.to_csv(
    "dataset_cleaned/ai_industry_summary_clean.csv",
    index=False
)

print("Cleaned datasets exported successfully.")
```
